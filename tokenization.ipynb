#2.2 Tokenizing Text

```python
#We are storing our training data in raw_text file
with open("Ali Baba and the Forty Thieves.md", "r",encoding="utf-8") as f:
    raw_text=f.read()
```

```python
#First we are converting the text/data into tokens
import re

text  = "Hello World.This is @ test."
result=re.split(r'([,.><?/;:!@#$%^&*()]|--|\s)', raw_text) #re.split is a function to split words at some chars
preprocessed = [item.strip() for item in result if item.strip()] #removing spaces

len(preprocessed)
```

#2.3 Converting tokens into token IDs 

```python
#making a collection of unique tokens
uniqueset=sorted(set(preprocessed)) #remove duplicate tokens and sort;it is used to build vocab
# len(uniqueset)
```

```python
#mapping each token of the uniqueset to an integer value
vocab = {token: integer for integer,token in enumerate(uniqueset)}
vocabsize=len(uniqueset)
# vocab
```

```python
class SimpleTokenizerV1:
    def __init__(self,vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}

    def encode(self,text):
        preprocessed = re.split(r'([,.><?/;:!@#$%^&*()]|--|\s)', text)
        preprocessed = [item.strip() for item in preprocessed if item.strip()]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids

    def decode(self,ids):
        text=" ".join(self.int_to_str[i] for i in ids)
        #keep spaces before the specified punctuations
        text=re.sub(r'\s + ([,.><?/;:!@#$%^&*()\'])',r'\1',text)
        return text
```

```python
tokenizer=SimpleTokenizerV1(vocab)

ids= tokenizer.encode(raw_text)
print(ids)

decoded_text=tokenizer.decode(ids)
print(decoded_text)
```
